# Preparación de los Datos

Antes de aplicar algoritmos de Aprendizaje Automático es necesario normalizar las variables para evitar que diferencias de escala afecten los resultados.

```python
X = df[[
    "AGE20X",
    "cantidad_condiciones",
    "recetas_totales",
    "benzodiacepinas",
    "antidepresivos",
    "hipnoticos",
    "ansioliticos",
    "total_psicofarmacos"
]]

print("Columnas seleccionadas:", X.shape)
print(X.dtypes)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

# Reducción de Dimensionalidad mediante PCA

Se utiliza Principal Component Analysis (PCA) para reducir la dimensionalidad y facilitar la visualización de los datos.

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
```

```python
varianza = pd.DataFrame({
    "Componente": [
        "PC1",
        "PC2"
    ],
    "Varianza explicada":
    pca.explained_variance_ratio_
})

varianza
```

# Selección del Número de Clusters

Se utiliza el Método del Codo para determinar la cantidad adecuada de grupos.

```python
inercias = []

for k in range(1,11):

    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    modelo.fit(X_pca)

    inercias.append(
        modelo.inertia_
    )
```

```python
pca_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
display(pca_df.head())
```

```python
inertia = []
K = range(1, 10)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(K, inertia, 'o-', color="#E63946", linewidth=2, markersize=8)
plt.title("Método del Codo (Elbow Method)", fontsize=16, fontweight='bold')
plt.xlabel("Número de Clusters (k)", fontsize=12)
plt.ylabel("Inercia", fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

# Modelo de Clustering K-Means

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_pca)

df["cluster"] = clusters


print("KMeans completado!")
print("Distribución de clusters:")
print(df["cluster"].value_counts())

display(df[["AGE20X", "total_psicofarmacos", "cluster"]].head(10))


print("\nPromedios por cluster:")
display(df.groupby("cluster").mean(numeric_only=True).round(2))
```

# Evaluación del Modelo

```python
silhouette = silhouette_score(
    X_pca,
    clusters
)

pd.DataFrame({
    "Métrica":[
        "Silhouette Score"
    ],
    "Valor":[
        round(
            silhouette,
            4
        )
    ]
})
```

# Visualización de Clusters

```python
plt.figure(figsize=(10, 8))

colors = ['#E63946', '#F77F00', '#6A0DAD']   # Rojo, Naranja, Morado

for i in range(3):
    plt.scatter(
        X_pca[df["cluster"] == i, 0],
        X_pca[df["cluster"] == i, 1],
        s=50,
        color=colors[i],
        label=f'Cluster {i}',
        alpha=0.85,
        edgecolors='black',
        linewidth=0.5
    )

plt.title("Visualización de Clusters en el Espacio PCA", fontsize=16, fontweight='bold')
plt.xlabel("Componente Principal 1", fontsize=12)
plt.ylabel("Componente Principal 2", fontsize=12)
plt.legend(title="Clusters", fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()
```

# Conclusiones

La aplicación de técnicas de Aprendizaje Automático no supervisado en este estudio permitió identificar de forma clara perfiles diferenciados de consumo de psicofármacos en una población de 27.805 individuos analizados.

Mediante el Análisis de Componentes Principales (PCA) se redujo exitosamente la dimensionalidad del conjunto de datos, reteniendo un 56.89% de la varianza original en solo dos componentes principales. Esta reducción facilitó la visualización de los patrones subyacentes y preparó los datos de manera óptima para la etapa de clustering.

Posteriormente, el algoritmo K-Means con 3 clusters segmentó a los individuos en grupos homogéneos según sus características demográficas (principalmente AGE20X), clínicas (cantidad_condiciones) y farmacológicas (recetas_totales, benzodiacepinas, antidepresivos, hipnoticos, ansioliticos y total_psicofarmacos).

Los resultados obtenidos confirman la existencia de patrones bien definidos de consumo de psicofármacos, respaldando plenamente la hipótesis inicial del proyecto. La combinación de PCA y K-Means demostró ser una aproximación eficaz y robusta para descubrir estructuras ocultas en datos de salud pública.
