# Modelos a implementar

En esta situacion vamos a implementar un modelo de aprendizaje supervisado con una tecnica de simplificacion y otra sin simplificacion, luego se comparan para saber cual queda mejor adaptada para el problema a partir de métricas y definiciones que se determinarán durante el desarrollo.

# Preparamos el dataset
Debido a que tenemos 4 features como entrada o datos caracteristicos, se aplica la tecnica PCA que reduce las dimensiones de 4 a 2, de esa forma nos ayuda a obener una grafica en un plano cartesiano x,y.

```python
#Importacion de las librerias necesarias
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from matplotlib.colors import ListedColormap
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
#Cargamos el dataframe procesado de nuestro repositorio de GitHub
url = 'https://raw.githubusercontent.com/Axumis/Parcial_Aprendizaje_Automatico/refs/heads/main/data/processed/dataset_con_riesgo.csv'
df = pd.read_csv(url)

#Mostramos el dataframe
df
```

# 1. K-Vecinos Cercanos (con PCA)

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.decomposition import PCA
X= df[['Temperatura_[°C]', 'Humedad_relativa_[%]', 'Precipitación_[mm]', 'Velocidad_viento_[km/h]']].values
y= df['Riesgo'].values

# 1. Codificamos las etiquetas
le = LabelEncoder()
y_encoded = le.fit_transform(y)

#2. Dividimos los datos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, random_state=0)
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 3. Escalamos las features tanto de entrenamiento como de prueba
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Aplicamos PCA
pca = PCA(n_components=2) # Reducimos a 2 dimensiones para el grafico
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)
```

Previo a la reducción de dimensiones, aplicamos una normalizacion puesto a que las variables estan en diferentes ordenes de magnitud y debido a que este modelo es bastante suceptible a sesgos o incertidumbre en el entrenamiento por parte de las dimensiones en los datos se aplica el metodo de MinMaxScaler()

# Entrenamos el modelo
Para esta situacion se aplica un valor de ***K = 4*** y para obtener un resultado del entrenamiento se confecciona un grafico que indica las clasificaciones de cada etiqueta a la que el modelo considera es la adecuada.

```python
# 5. Entrenamos y predecimos
k = 4 #Asignamos un valor de k
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train_pca, y_train)
```

El resultado del computo para este entrenamiento resulto ser de:

image.png

Vamos a comparar mas adelante con el siguiente modelo

# Gráfica de la Clasificación Obtenida

Visualizamos el entrenamiento (Utilizando PCA para reducir dimensiones)

```python
# Crear malla para frontera de decisión
h = 0.5
x_min, x_max = X_train_pca[:, 0].min() - 1, X_train_pca[:, 0].max() + 1
y_min, y_max = X_train_pca[:, 1].min() - 1, X_train_pca[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_min, y_max, h))
Z = knn.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Colores
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF', '#FFFFAA', '#FFAACC'])
cmap_bold = ListedColormap(['#FF0000', '#00FF00', '#0000FF', '#CCCC00', '#CC00CC'])

# Graficar
plt.figure(figsize=(10, 6))
plt.contourf(xx, yy, Z, cmap=cmap_light, alpha=0.6)
scatter = plt.scatter(X_train_pca[:, 0], X_train_pca[:, 1], c=y_train,
                      cmap=cmap_bold, edgecolor='k', s=40)

# Crear leyenda con los nombres originales de las clases
legend_labels = le.inverse_transform(np.unique(y_train))
legend_colors = [scatter.cmap(scatter.norm(i)) for i in np.unique(y_train)]
patches = [plt.Line2D([0], [0], marker='o', color='w',
                      label=label, markerfacecolor=color, markersize=10)
           for label, color in zip(legend_labels, legend_colors)]

plt.legend(handles=patches, title="Clases")
plt.title(f"Visualización de KNN con PCA (k = {k}, {len(legend_labels)} clases)")
plt.xlabel("Componente principal 1")
plt.ylabel("Componente principal 2")
plt.grid(True)
plt.tight_layout()
plt.show()
```

Se observa un grafico a la cuál dificulta la lectura de las clasificaciones, puesto a que se tienen, en primer lugar, los datos estan muy agrupados, es decir, muy juntos, lo cual dificulta observar una linea o frontera que separe los datos de forma precisa. La otra cuestion es la reduccion de las dimensiones a 2 aplicando PCA, debido a esto, el grafico no aporta mucha informacón en cuanto a la clasifciacion de este problema. Para ello se procede con observar las métricas.

# Métricas del modelo

```python
# Predicciones
y_pred = knn.predict(X_test_pca)

# Accuracy
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc:.2f}")

# Reporte completo
print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred, target_names=le.inverse_transform(np.unique(y_encoded))))
```

Se tiene dentro de todo una buena metrica dentro de todo, con un accuracy del 93% de acertación de los datos, considerando las controversias en las reducciones de las diemensiones. Ademeás las otras metricas dan unos resultados altos que se pueden apreciar mejor en el grafico de barras del codigo debajo.

```python
# Obtener el reporte de clasificación como diccionario
report = classification_report( y_test, y_pred, target_names=le.inverse_transform(np.unique(y_encoded)), output_dict=True)

# Convertir a DataFrame y eliminar la fila 'accuracy', 'macro avg' y 'weighted avg'
df_report = pd.DataFrame(report).transpose()
df_clases = df_report.iloc[:-3]  # solo las clases

# Plot
plt.figure(figsize=(10, 6))
df_clases[['precision', 'recall', 'f1-score']].plot(kind='bar', figsize=(10, 6), colormap='Set2')

plt.title("Métricas por clase (Precisión, Recall, F1-Score)")
plt.xlabel("Clase")
plt.ylabel("Valor")
plt.ylim(0, 1.05)
plt.grid(axis='y')
plt.xticks(rotation=45)
plt.tight_layout()
plt.legend(loc='lower right')
plt.show()
```

Se puede observar que este entrenamiento da buenos nuemeros en las metricas de precision, recall y f1-score en las diferentes categorias que considera este modelo clasificar. Con el valor más minimo de todas ellas la que corresponde a la etiqueta **"Extremo".** Esto es debido a que posee pocos datos de entrenamiento y de prueba. Otra forma mejor de ver los resultados a este modelo es observando la matriz de confución.

```python
# Matriz de confusión
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=le.inverse_transform(np.unique(y_encoded)),
            yticklabels=le.inverse_transform(np.unique(y_encoded)))
plt.xlabel("Predicción")
plt.ylabel("Clase real")
plt.title("Matriz de confusión")
plt.tight_layout()
plt.show()
```

La matriz arroja claramente los aciertos y desaciertos en las predicciones de los datos. Con muchos fallos en las categorias "Bajo" y "Moderado" (en el orden de los 100) Mientras que las demás van entre muchos aciertos y pocas predicciones fallidas (entre 30 y 40 fallas). Se puede considerar a las clases "Bajo" Y "Moderado" como las mas concentradas en datos pueso a que la regla asigno más de esos datos a esas categorias lo que indica que puede haber un mayor grado de tolerancia al error para esas dos etiquetas. Aunque lo que se busca es reducir los errores a un grado minimo. Pero vamos a contrastar estas mismas condiciones al siguiente modelo pero sin utilizar PCA.

#2. Modelo K-NN (sin usar PCA)

Realizamos otro modelo identico al anterior con la diferencia que no aplicamos la tecnica de reduccion de variables, para compararlas y ver cual se ajusta mejor al problema

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.decomposition import PCA
X= df[['Temperatura_[°C]', 'Humedad_relativa_[%]', 'Precipitación_[mm]', 'Velocidad_viento_[km/h]']].values
y= df['Riesgo'].values


#1. Dividimos los datos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
scaler = MinMaxScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 3. Escalamos las features tanto de entrenamiento como de prueba
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Entrenamos el modelo

Le asignamos el mismo valor de ***K=4*** que el primer modelo

```python
# 5. Entrenamos y predecimos
k = 4 #Asignamos un valor de k
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train, y_train)
```

El resultado de la ejecucion para esta situacion se determino en:

image.png

Lo que indica que no hay mucha diferencia

# Metricas del modelo

Aplicamos las predicciones y realizamos las metricas que ayuden a decidir en que modelo se ajusta mejor

```python
# Predicciones
y_pred = knn.predict(X_test)

# Accuracy
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc:.2f}")

# Reporte completo
print("\nReporte de clasificación:")
clases = ['Bajo', 'Moderado', 'Alto', 'Muy alto', 'Extremo']
print(classification_report(y_test, y_pred, labels=clases, target_names=clases))
```

Hay una mejoria en el accuracy del 96% pero no mucha diferencia respecto del modelo anterior con respecto a las metricas, pero aun asi, este modelo indica mejorias y mas precision.

```python
# Obtener el reporte de clasificación como diccionario
report = classification_report(y_test, y_pred, output_dict=True)

# Convertir a DataFrame y eliminar la fila 'accuracy', 'macro avg' y 'weighted avg'
df_report = pd.DataFrame(report).transpose()
df_clases = df_report.iloc[:-3]  # solo las clases

# Plot
plt.figure(figsize=(10, 6))
df_clases[['precision', 'recall', 'f1-score']].plot(kind='bar', figsize=(10, 6), colormap='Set2')

plt.title("Métricas por clase (Precisión, Recall, F1-Score)")
plt.xlabel("Clase")
plt.ylabel("Valor")
plt.ylim(0, 1.05)
plt.grid(axis='y')
plt.xticks(rotation=45)
plt.tight_layout()
plt.legend(loc='lower right')
plt.show()
```

Desde el grafico se puede distinguir muy bien las metricas, obervamos que todas las categorias tienen un porcentaje superior al 90% de igual forma que el modelo anterior, pero con una leve diferencia en mejorias de metricas en éste, lo cual indica un excelente modelo para implementar y predecir los niveles de riesgos

```python
# Calcular la matriz de confusión
cm = confusion_matrix(y_test, y_pred)

# Obtener etiquetas de clase únicas ordenadas
labels = sorted(np.unique(y_test))

# Graficar la matriz
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=labels,
            yticklabels=labels)

plt.xlabel("Predicción")
plt.ylabel("Clase real")
plt.title("Matriz de confusión")
plt.tight_layout()
plt.show()
```

Se puede ver claramente desde la matriz como los errores son mucho mas reducidos. Esto nos dice, que este modelo se ajusta mejor a las predicciones porque justamente este modelo no utiliza un metodo de reuducción en las dimensiones de las features por lo que se considera al conjunto de forma. Aunque se sabe que al no reducir el tamaño en dimensiones hace que el nivel de computo sea mas elevado, para este problema, la diferencia es muy infimo, por lo que se puede adaptar tranquilamente este modelo ya para cualquier valor de K.
